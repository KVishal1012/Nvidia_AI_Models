```python
import os 
import getpass

os.environ["NVIDIA_API_KEY"] = getpass.getpass("Enter your NVIDIA API Key: ")
```

```python
from openai import OpenAI

client = OpenAI(
    base_url = "https://integrate.api.nvidia.com/v1",
    api_key = os.environ["NVIDIA_API_KEY"]  
)

print("NVIDIA API is working correctly!")
```

```python
models = client.models.list()

for model in models.data[:10]: # Print only the first 10 models
    print(f"Model ID: {model.id}, Created: {model.created}")
```

```python
models = client.models.list()

for model in models.data:
    if "nemotron" in model.id.lower() or "nvidia" in model.id.lower():
        print(model.id)
```

```python
available_models = {model.id for model in client.models.list().data}

preferred_models = [
    "nvidia/cosmos-reason2-8b",
    "nvidia/cosmos-reason2-2b",
    "nvidia/cosmos3-nano-reasoner"
]

MODEL = None

for candidate in preferred_models:
    if candidate in available_models:
        MODEL = candidate
        break

if MODEL is None:
    print("No preferred Cosmos model found. Available Cosmos models:")
    for model_id in sorted(available_models):
        if "cosmos" in model_id.lower():
            print(model_id)
else:
    print("Selected model:", MODEL)
```

```python
response = client.chat.completions.create(
    model=MODEL,
    messages=[
        {"role": "system", "content": "You are a GeoAI assistant focused on physical-world reasoning, urban infrastructure and flood monitoring."},
        {"role": "user", "content": """
Create a physical-world reasoning workflow for urban flood monitoring in Chennai, India.
         
Use:
- Sentinel-1 SAR
- Sentinel-2 imagery
- rainfall data
- roads
- buildings
- drainage network
         
Explain
1. What the model should observe
2. What physical-world reasoning is needed
3. What GIS rules should be deterministic
4. What the final output should look like as JSON
        """
         }
    ],
    temperature=0.2,
    max_tokens=1200
)

print(response.choices[0].message.content)
```
