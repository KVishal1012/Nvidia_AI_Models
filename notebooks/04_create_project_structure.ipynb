```python
from pathlib import Path

project_root = Path("/content/drive/MyDrive/chennai-flood-mvp")

folders = [
    "notebooks",
    "src",
    "data/raw",
    "data/processed",
    "data/outputs",
    "docs"
]

for folder in folders:
    (project_root / folder).mkdir(parents=True, exist_ok=True)

starter_files = {
    "src/config.py": "# Project configuration\n",
    "src/download_data.py": "# Download Sentinel, rainfall, roads, buildings, and drainage data\n",
    "src/flood_mask.py": "# Rule-based Sentinel-1 SAR flood mask generation\n",
    "src/optical_extent.py": "# Sentinel-2 NDWI/MNDWI flood extent mapping\n",
    "src/rainfall.py": "# Rainfall threshold and simple flood prediction logic\n",
    "src/accessibility.py": "# Road accessibility analysis using flooded road segments\n",
    "src/exposure.py": "# Building exposure and simple damage assessment\n",
    "src/drainage_risk.py": "# Simple drainage risk analysis\n",
    "src/export_json.py": "# Export final flood risk output as JSON/GeoJSON\n",
    "requirements.txt": "",
    "README.md": "# Chennai Flood MVP\n\nBeginner-friendly GeoAI flood monitoring project for Chennai, India.\n"
}

for file_path, content in starter_files.items():
    path = project_root / file_path
    if not path.exists():
        path.write_text(content, encoding="utf-8")

print(f"Project created at: {project_root}")
```
